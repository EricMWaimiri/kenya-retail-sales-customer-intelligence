## Convert Date Columns to DT 

```python
# Convert all date columns to pandas datetime format
date_columns = {
    "sales": (sales, ["TransactionDate"]),
    "customers": (customers, ["signup_date"]),
    "promotions": (promotions, ["StartDate", "EndDate"]),
    "holidays": (holidays, ["date", "observed_date"]),
    "cpi": (cpi, ["date"]) 
}

for name, (df, columns) in date_columns.items():
    for col in columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")
    print(f"{name}: date conversion complete")

# Verify the converted date columns
for name, (df, columns) in date_columns.items():
    print(f"\n{name.upper()}")
    print(df[columns].dtypes)
```

## Standardize Column Names

```python
def to_pascal(name: str) -> str:
    # Insert space before capitals that follow lowercase/digit, then title-case each word
    name = re.sub(r'(?<=[a-z0-9])([A-Z])', r' \1', name)
    name = name.replace("-", " ").replace("_", " ")
    return "".join(word.capitalize() for word in name.split())

def std_columns(df):
    df = df.copy()
    df.columns = [to_pascal(col) for col in df.columns]
    return df

# Apply the function to required datasets
customers = std_columns(customers)
holidays = std_columns(holidays)
cpi = std_columns(cpi)

# Display the standardized column names
for name, df in {
    "Sales": sales,
    "Customers": customers,
    "Products": products,
    "Stores": stores,
    "Promotions": promotions,
    "Public Holidays": holidays,
    "KNBS CPI": cpi
}.items():
    print(f"\n{name} columns:")
    print(df.columns.tolist())
```

## Outlier check (IQR rule)

Outliers are checked on the three numeric columns where an extreme value would matter: `Quantity`, `UnitPriceKsh` and `DiscountPct`. A value is flagged when it lies more than 1.5 × IQR outside Q1–Q3.

```python
def iqr_check(series, name):
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr = q3 - q1
    lower, upper = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    flagged = ((series < lower) | (series > upper)).sum()
    return {"Column": name, "Min": series.min(), "Q1": q1, "Q3": q3, "Max": series.max(),
            "Lower fence": lower, "Upper fence": upper, "Outliers": int(flagged)}

outlier_table = pd.DataFrame([
    iqr_check(sales["Quantity"], "sales.Quantity"),
    iqr_check(products["UnitPriceKsh"], "products.UnitPriceKsh"),
    iqr_check(sales["DiscountPct"], "sales.DiscountPct"),
])
display(outlier_table)
print(sales["Quantity"].value_counts().sort_index().to_dict())
print(sales["DiscountPct"].value_counts().sort_index().to_dict())
```

**Decision: keep every flagged row.**

- `Quantity` (620 flagged): baskets of 7 to 10 units are plausible and the maximum is only 10, so none are entry errors.
- `UnitPriceKsh` (9 flagged): the flagged products are genuinely expensive items (the highest is KSh 11,500, a household appliance). Removing them would delete the products that drive most of the sales.
- `DiscountPct` (2,500 flagged): 12,500 of 15,000 lines have no discount, so the IQR fence is 0 and every discounted line is flagged. Discounts range from about 4% to 25%, which is a normal promotional range.

Nothing is dropped, because each flagged value is a real business value. The tests in Notebook 04 use rank-based methods, which are robust to these values.

## Feature engineering

Derived columns are built here and saved to a **new** file (`sales_features.parquet`), so `sales_cleaned` stays exactly as the later notebooks and the Power BI model expect.

```python
df = sales.merge(products[["ProductID", "Category", "UnitCostKsh", "UnitPriceKsh"]], on="ProductID", how="left")
df["SaleDate"] = df["TransactionDate"].dt.normalize()
df["SalesValue"] = df["Quantity"] * df["UnitPriceKsh"]                 # list price x quantity
df["NetSales"] = df["SalesValue"] * (1 - df["DiscountPct"] / 100)       # after discount
df["TotalCost"] = df["Quantity"] * df["UnitCostKsh"]
df["GrossProfit"] = df["NetSales"] - df["TotalCost"]                    # estimated
df["Month"] = df["TransactionDate"].dt.to_period("M").dt.to_timestamp()
df["Weekday"] = df["TransactionDate"].dt.day_name()

holiday_dates = set(pd.to_datetime(holidays["ObservedDate"]).dt.normalize())
df["IsHoliday"] = df["SaleDate"].isin(holiday_dates)

promo_dates = set()
for _, p in promotions.iterrows():
    promo_dates.update(pd.date_range(p["StartDate"], p["EndDate"]).normalize())
df["HasPromotion"] = df["SaleDate"].isin(promo_dates)

df = df.merge(customers[["Id", "SignupDate"]], left_on="CustomerID", right_on="Id", how="left").drop(columns="Id")
df["BeforeSignup"] = df["SaleDate"] < df["SignupDate"]

print("Shape:", df.shape)
print("Holiday lines:", int(df["IsHoliday"].sum()))
print("Promotion lines:", int(df["HasPromotion"].sum()))
print("Lines before signup date:", int(df["BeforeSignup"].sum()),
      "from", df.loc[df["BeforeSignup"], "CustomerID"].nunique(), "customers")
print("Share of net sales flagged:", round(df.loc[df["BeforeSignup"], "NetSales"].sum() / df["NetSales"].sum() * 100, 1), "%")

df.to_parquet("../data/cleaned/sales_features.parquet", index=False)
df.to_csv("../data/cleaned/sales_features.csv", index=False)
```

**Purchases before signup date.** Some lines are dated before the customer's `SignupDate`. These are **kept and flagged** (`BeforeSignup`), not deleted, because `SignupDate` may be a loyalty-registration date rather than a first-purchase date. The flag lets any analysis exclude them if needed.
