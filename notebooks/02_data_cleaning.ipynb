## Cleaning the Data

```python
import pandas as pd
import numpy as np
import re
from IPython.display import display
```

```python
sales = pd.read_csv("../data/raw/synthetic/sales.csv")
customers = pd.read_csv("../data/raw/synthetic/customers.csv")
products = pd.read_csv("../data/raw/synthetic/products.csv")
stores = pd.read_csv("../data/raw/synthetic/stores.csv")
promotions = pd.read_csv("../data/raw/synthetic/promotions.csv")
holidays = pd.read_csv("../data/raw/synthetic/public_holidays.csv")
cpi = pd.read_csv("../data/raw/knbs/national_cpi_monthly.csv")
```

```python

# Basic DataFrame Checks Function
def basic_df_checks(df_checks):

    print("=== BASIC DATAFRAME CHECKS ===")
    print(f"Shape: {df_checks.shape[0]} rows × {df_checks.shape[1]} columns\n")

    print("Column Data Types:")
    print(df_checks.dtypes)

    print("\nMissing Values:")
    print(df_checks.isnull().sum())

    print("\nDuplicate Rows:", df_checks.duplicated().sum())

    print("\n=== NUMERIC SUMMARY ===")
    numeric_cols = df_checks.select_dtypes(include=np.number).columns

    if len(numeric_cols) > 0:
        display(df_checks[numeric_cols].describe())
    else:
        print("No numeric columns found.")

    print("\n=== CATEGORICAL SUMMARY ===")
    categorical_cols = df_checks.select_dtypes(
        include=["object", "category", "string"]
    ).columns

    if len(categorical_cols) > 0:
        display(df_checks[categorical_cols].describe().T)
    else:
        print("No categorical columns found.")

    print("\nUnique Values per Column:")
    for col in df_checks.columns:
        print(f"{col}: {df_checks[col].nunique(dropna=True)}")

# Run the basic checks on the sales DataFrame
basic_df_checks(sales)
```

## Convert Date Columns to DT 

```python
# Convert all date columns to pandas datetime format
date_columns = {
    "sales": (sales, ["TransactionDate"]),
    "customers": (customers, ["signup_date"]),
    "promotions": (promotions, ["StartDate", "EndDate"]),
    "holidays": (holidays, ["date", "observed_date"]),
    "cpi": (cpi, ["date"]) 
}

for name, (df, columns) in date_columns.items():
    for col in columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")
    print(f"{name}: date conversion complete")

# Verify the converted date columns
for name, (df, columns) in date_columns.items():
    print(f"\n{name.upper()}")
    print(df[columns].dtypes)
```

## Standardize Column Names

```python
def to_pascal(name: str) -> str:
    # Insert space before capitals that follow lowercase/digit, then title-case each word
    name = re.sub(r'(?<=[a-z0-9])([A-Z])', r' \1', name)
    name = name.replace("-", " ").replace("_", " ")
    return "".join(word.capitalize() for word in name.split())

def std_columns(df):
    df = df.copy()
    df.columns = [to_pascal(col) for col in df.columns]
    return df

# Apply the function to required datasets
customers = std_columns(customers)
holidays = std_columns(holidays)
cpi = std_columns(cpi)

# Display the standardized column names
for name, df in {
    "Sales": sales,
    "Customers": customers,
    "Products": products,
    "Stores": stores,
    "Promotions": promotions,
    "Public Holidays": holidays,
    "KNBS CPI": cpi
}.items():
    print(f"\n{name} columns:")
    print(df.columns.tolist())
```

## Resolve Missing Values

```python
# Drop the promotion ID column from the sales dataset
sales = sales.drop(columns=["PromotionID"])

# Confirm the column has been removed
print("Updated sales columns:")
print(sales.columns.tolist())
```

```python

# Fill missing observed dates with the actual holiday dates assuming the holiday was observed on its listed date.
holidays["ObservedDate"] = holidays["ObservedDate"].fillna(holidays["Date"])
print(holidays["ObservedDate"].isnull().sum())

# Display the holiday dates and observed dates for verification
display(
    holidays[["Date", "HolidayName", "ObservedDate"]].head(10)
)
```

## Correct Spelling Inconsistency

```python
customers["CustomerSegment"] = customers["CustomerSegment"].replace(
    "Occassional", "Occasional")

print(customers["CustomerSegment"].value_counts())
```

```python
# Prefer Parquet (keeps datetime, categories, etc.)
sales.to_parquet("../data/cleaned/sales_cleaned.parquet", index=False)
customers.to_parquet("../data/cleaned/customers_cleaned.parquet", index=False)
products.to_parquet("../data/cleaned/products_cleaned.parquet", index=False)
stores.to_parquet("../data/cleaned/stores_cleaned.parquet", index=False)
holidays.to_parquet("../data/cleaned/public_holidays_cleaned.parquet", index=False)
promotions.to_parquet("../data/cleaned/promotions_cleaned.parquet", index=False)
cpi.to_parquet("../data/cleaned/national_cpi_monthly_cleaned.parquet", index=False)

print("All cleaned datasets have been saved!")
```

## Data Cleaning Summary

- Standardized column names.
- Converted `TransactionDate` in sales, `signup_date` in customers, `StartDate` and `EndDate` in promotions, and `Date` and `ObservedDate` in holidays to datetime format.
- Dropped the `PromotionID` column from the sales dataset.
- Filled missing `ObservedDate` values using `Date`.
- Corrected the customer segment spelling from `Occassional` to `Occasional`.
- Saved all seven datasets to the `data/cleaned/` folder.
