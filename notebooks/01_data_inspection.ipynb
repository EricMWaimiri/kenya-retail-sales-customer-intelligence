```python
from pathlib import Path
import pandas as pd

RAW = Path("D:/capstone/data/raw/synthetic")

files = [
    "sales.csv",
    "customers.csv",
    "products.csv",
    "stores.csv",
    "promotions.csv",
    "public_holidays.csv",
]

for csv in files:
    path = RAW / csv

    print(f"\n{'=' * 55}")
    print(f"FILE: {csv}")

    df = pd.read_csv(path)

    
    print("Number of rows:", len(df))

    print("Number of columns:", len(df.columns))

    print("Column names:")
    print(df.columns)

    print("First 3 rows:")
    print(df.head(3))

    print("Missing values in each column:")
    print(df.isnull().sum())

```

```python
# Load CSV files
sales = pd.read_csv("../data/raw/synthetic/sales.csv")
customers = pd.read_csv("../data/raw/synthetic/customers.csv")
products = pd.read_csv("../data/raw/synthetic/products.csv")
stores = pd.read_csv("../data/raw/synthetic/stores.csv")
promotions = pd.read_csv("../data/raw/synthetic/promotions.csv")
holidays = pd.read_csv("../data/raw/synthetic/public_holidays.csv")
cpi = pd.read_csv("../data/raw/knbs/national_cpi_monthly.csv")

print("DATA QUALITY CHECK")
print("==================")

# Check for duplicate SaleIDs
print("\nDuplicate SaleIDs:")
print(sales["SaleID"].duplicated().sum())

# Check for invalid quantities
print("\nSales with quantity zero or less:")
print((sales["Quantity"] <= 0).sum())

# Check discount values
print("\nDiscounts below 0 or above 100:")
print(((sales["DiscountPct"] < 0) |
       (sales["DiscountPct"] > 100)).sum())

# Check sales dates
sales["TransactionDate"] = pd.to_datetime(
    sales["TransactionDate"], errors="coerce"
)

print("\nInvalid sales dates:")
print(sales["TransactionDate"].isna().sum())

print("\nEarliest sales date:")
print(sales["TransactionDate"].min())

print("\nLatest sales date:")
print(sales["TransactionDate"].max())

# Check customer IDs
print("\nSales with unknown customers:")
print((~sales["CustomerID"].isin(customers["id"])).sum())

# Check product IDs
print("\nSales with unknown products:")
print((~sales["ProductID"].isin(products["ProductID"])).sum())

# Check store IDs
print("\nSales with unknown stores:")
print((~sales["StoreID"].isin(stores["StoreID"])).sum())

# Check promotion IDs
sales_promotions = sales["PromotionID"].dropna()
known_promotions = promotions["PromotionID"]

print("\nSales with unknown promotion IDs:")
print((~sales_promotions.isin(known_promotions)).sum())

# Check customer segment spelling
print("\nCustomer segment counts:")
print(customers["customer_segment"].value_counts())

print("\nChecks completed.")
```
